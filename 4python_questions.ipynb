```python
# Q Count how many times a number appears in a list:
nums = []
n = int(input("Enter number of elements: "))

for i in range(n):
    x = int(input("Enter number: "))
    nums.append(x)

target = int(input("Enter number to count: "))

count = 0
for x in nums:
    if x == target:
        count += 1

print("Count:", count)


```

```python
# Find the sum of all elements in a list (without using sum()).

nums = []
n = int(input("Enter number of elements: "))

for i in range(n):
    x = int(input("Enter number: "))
    nums.append(x)

total = 0
for x in nums:
    total += x

print("Sum:", total)
```

```python
# Reverse a list using a loop (don’t use reverse() function).
nums = []
n = int(input("Enter number of elements: "))

for i in range(n):
    nums.append(int(input("Enter number: ")))

start = 0
end = len(nums) - 1

while start < end:
    nums[start], nums[end] = nums[end], nums[start]
    start += 1
    end -= 1

print("Reversed list:", nums)

```

```python
nums = [1, 2, 3, 4, 5]

start = 0                 # starting index
end = len(nums) - 1       # last index

while start < end:        # loop until pointers meet
    nums[start], nums[end] = nums[end], nums[start]  # swap
    start += 1            # move start forward
    end -= 1              # move end backward

print(nums)
```

```python
# Remove all occurrences of "apple" from a list.
fruits = ["apple", "banana", "apple", "orange", "apple", "grapes"]

new_list = []
for x in fruits:
    if x != "apple":
        new_list.append(x)

print(new_list)
```
