## Pie Chart
    Used to show percentage distribution of a whole.

```python
labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 30, 15, 15]
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen']

plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=360)

plt.title("Programming Language Popularity")
plt.axis('equal')  # Equal aspect ratio ensures pie is circular
plt.show()
```


    autopct='%1.1f%%': show % labels.

    startangle=140: rotates start angle.

    plt.axis('equal'): ensures perfect circle.

# Subplots – Multiple Charts in One Figure
    Used when you want to show multiple plots side by side for comparison.

### Syntax: plt.subplot(rows, columns, index)

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [25, 16, 9, 4, 1]

plt.figure(figsize=(10, 4))

# First plot
plt.subplot(2, 2, 1)  # 1 row, 2 columns, 1st plot
plt.plot(x, y1, color='green')
plt.title("Square Numbers")

# Second plot
plt.subplot(2, 2, 3)  # 1 row, 2 columns, 2nd plot
plt.plot(x, y2, color='red')
plt.title("Reverse Squares")

plt.tight_layout()
plt.show()
```

    subplot(1, 2, 1): First plot in a 1x2 grid.

    tight_layout(): prevents overlap between subplots.



## Annotations – Highlighting Points

    Use Case:
    To label or point out a specific value or point of interest.

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y, marker='o')
plt.title("Line Chart with Annotation")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

# Annotate the last point
plt.annotate('Max Point', xy=(5, 10), xytext=(3, 9),
             arrowprops=dict(facecolor='black', arrowstyle='->'))

plt.grid(True)
plt.show()
```

# Interpolation in Matplotlib

    Interpolation is a method of estimating unknown values between two known values. In plotting, it helps to create a smooth curve from a small set of data points.



    Basic Interpolation with scipy.interpolate
    To demonstrate interpolation effectively in matplotlib, we use the scipy library.

## Linear vs Interpolated (Smooth) Curve

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import interp1d

# Original data points
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([0, 0.8, 0.9, 0.1, -0.8, -1])

# Interpolation function - 'cubic' creates smooth curves
f_cubic = interp1d(x, y, kind='cubic')
f_linear = interp1d(x, y, kind='linear')

# New x values for smooth curve
x_new = np.linspace(0, 5, 100)
y_cubic = f_cubic(x_new)
y_linear = f_linear(x_new)

# Plot
plt.plot(x, y, 'o', label='Original Data')

plt.plot(x_new, y_cubic, '-', label='Cubic Interpolation')
plt.plot(x_new, y_linear, '--', label='Linear Interpolation')

plt.title("Interpolation Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid(True)
plt.show()
```

    Types of Interpolation:
    'linear': Straight-line connection between points.

    'quadratic': Fits a 2nd-degree polynomial.

    'cubic': Smoothest curve, fits a 3rd-degree polynomial.
