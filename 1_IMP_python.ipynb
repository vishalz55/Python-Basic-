```python
# 1️⃣ h = a ** b

# 👉 Exponentiation (power)

# It means:
# a raised to the power b

# Example:

# a = 2
# b = 3
# h = a ** b
# print(h)


# Output:

# 8


# 👉 Because: 2 × 2 × 2 = 8

# 2️⃣ i = a // b

# 👉 Floor division

# It means:
# Divide a by b and take only the integer part (no decimals)

# Example:

# a = 7
# b = 2
# i = a // b
# print(i)


# Output:

# 3


# 👉 Because: 7 / 2 = 3.5 → floor value = 3

# 🔍 Difference between / and //
# 7 / 2   → 3.5
# 7 // 2  → 3
```

```python

# Yes ✅ we can traverse a 2-D array using a for loop (actually, using nested for loops).
num = ["one","two","three"]
fruits = ["apple","banana","mango"]

for i in range(len(num)):
  for j in range(len(fruits)):
    print(num[i],fruits[j])

```
