# Numpy
NumPy is a popular Python library for numerical and scientific computing
Importing NumPy


```python
import numpy as np
```

NumPy Arrays
Numpy arrays can be created from lists, tuples, or other iterable objects.


```python
# Creating a 1D array
arr_1d = np.array([1, 2, 3, 4, 5])

arr_1d
```

```python
# Creating a 2D array
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
arr_2d
```

```python


# Creating arrays with specific values
zeros = np.zeros((2, 3))  # Array of zeros
print(zeros)

ones = np.ones((3, 4))  # Array of ones
print(ones)

identity = np.eye(3)  # Identity matrix
print(identity)
```

Array Operations
We can perform various operations on NumPy arrays, such as addition, subtraction, multiplication, and more.

```python
a = [1,2,3]
b = [3,4,6]
print(a+b)
```

```python

# Array operations
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

addition = a + b
subtraction = a - b
multiplication = a * b
division = a / b
```

Indexing and Slicing:
We can access elements and slices of arrays using indexing.

```python

# Indexing and slicing
arr = np.array([1, 2, 3, 4, 5])

element = arr[2]
slice = arr[1:4]
slice
```

Array Shape and Size:
We can check the shape and size of an array.

```python

# Shape and size
#shape = arr_2d.shape  # Shape of a 2D array
size = arr.size  # Total number of elements in an array
size
```

Reshaping Arrays:
We can change the shape of an array using the reshape method.

```python
# Reshaping arrays
arr = np.array([1, 2, 3, 4, 5, 6])
reshaped = arr.reshape(3, 2)
reshaped
```

Statistical Operations:
NumPy provides various statistical functions for analyzing data.

```python

# Statistical operations
mean = np.mean(arr)
median = np.median(arr)
std_dev = np.std(arr)
print(mean)
print(median)
print(std_dev)
```

Random Number Generation:
We can generate random numbers and arrays using NumPy.

```python
import random
# Random number generation
random_int = np.random.randint(1000, 50000)  # Random integer between 1 and 100
random_int
```

```python
random_array = np.random.rand(3, 4)  # 3x4 array of random floats
random_array
```

```python
# Generate a random floating-point number
random_number = np.random.uniform(0, 1)
print(random_number)


random_value = np.random.uniform(10, 20)
print(random_value)
```
